Flujo: descargar clima (NASA POWER) → calcular riesgo y ventana de aplicación → mirar el evento de O'Higgins 2026.

```python
# Permite importar desde src/ estando en notebooks/
import sys, pathlib
sys.path.append(str(pathlib.Path.cwd().parent))

from src.nasa_power import get_power_daily
from src import risk
import matplotlib.pyplot as plt
```

## 1. Descargar clima de la parcela

Ejemplo: zona de estudio anonimizada (centro de la celda de datos), primavera 2026.

```python
from src import anonimato
pt = anonimato.generalizar()
df = get_power_daily(lat=pt.lat, lon=pt.lon, start="20260901", end="20261031")
df.tail()
```

## 2. Riesgo de infección + ventana de aplicación

Sin humedad de suelo real todavía (SMAP se suma luego); por ahora la ventana usa solo lluvia.

```python
rec = risk.recomendacion(df)
rec.tail(15)
```

## 3. Visualizar el evento

Lluvia diaria + días de riesgo. Aquí se ve el pico tras las lluvias/granizo.

```python
fig, ax = plt.subplots(figsize=(11,4))
ax.bar(df.index, df["PRECTOTCORR"], width=0.8, label="Lluvia (mm)")
ax.scatter(rec.index[rec["riesgo"]==1], [0]*int(rec["riesgo"].sum()),
           color="crimson", marker="v", label="Día de riesgo")
ax.set_title("Lluvia y días de riesgo — zona piloto, primavera 2026")
ax.legend(); plt.tight_layout()
```
